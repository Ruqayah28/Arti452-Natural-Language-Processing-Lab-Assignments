## Task 1 — Canny Edge Detection

An **edge** is where brightness changes sharply. Canny finds edges in 5 steps: Gaussian blur → gradients → non-maximum suppression (thin the edges) → double thresholding → hysteresis (keep weak edges only if they connect to strong ones).

`cv2.Canny(img, T_low, T_high)`:
- `T_low` = lower threshold (weak edges)
- `T_high` = upper threshold (strong edges)
- Rule of thumb: `T_high ≈ 2×T_low` or `3×T_low`.

```python
# --- Code from the lab manual ---
img = data.camera()

edges = cv2.Canny(img, 100, 200)  #cv2.Canny(input_img, lower threshold, upped threshold)

plt.subplot(121), plt.imshow(img, cmap='gray')
plt.title('Original Image'), plt.xticks([]), plt.yticks([])
plt.subplot(122), plt.imshow(edges, cmap='gray')
plt.title('Edge Image'), plt.xticks([]), plt.yticks([])
plt.show()
```

## Task 2 — Harris Corner Detection

A **corner** is a point where the image changes in *two* different directions, which makes corners highly distinctive.



- Large positive `R` → corner
- Large negative `R` → edge
- Small `R` → flat region

Parameters: `blockSize` (window size), `ksize` (Sobel kernel size), `k` (sensitivity, typically 0.04–0.06).

> **Note:** OpenCV uses BGR order, so `(0, 0, 255)` is red.

```python
# --- Code from the lab manual ---
board = data.checkerboard()
image = cv2.cvtColor(board, cv2.COLOR_GRAY2BGR) #Convert from one channel to three channels, we need BGR to draw red dots
image = cv2.resize(image, (400, 400), interpolation=cv2.INTER_NEAREST)

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) #to analyze intensity changes we need grayscale image
gray = np.float32(gray) #floating-point data type expected by cv2.cornerHarris()

# Harris Corner Detection
harris = cv2.cornerHarris(gray, blockSize=2, ksize=3, k=0.04)

# Threshold for detecting strong corners
threshold = 0.01 * harris.max()
corner_coords = np.argwhere(harris > threshold) #np.argwhere() finds the coordinates of those True pixels.

# Draw larger red dots (radius = 7)
for y, x in corner_coords:
    cv2.circle(image, (x, y), radius=7, color=(0, 0, 255), thickness=-1)  # -1 = filled

# Display
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)) #converts BGR to RGB because Matplotlib expects RGB channel order.
plt.title('Harris Corners with Bigger Red Dots')
plt.axis('off')
plt.show()

'''Notes
Grayscale → BGR: Prepare the image for colored annotations.
BGR → Grayscale: Prepare the image for Harris corner detection.
np.float32(): Convert to the required numerical data type.'''
```

blockSize:Size of the local neighborhood used to calculate intensity changes.

ksize:Size of the Sobel kernel used to calculate image gradients.

k:Sensitivity constant used in the Harris response formula.

## Q1 — Otsu's Thresholding (Assessment Task)
Otsu picks `T` **automatically** by maximizing the between-class variance — assumes the histogram has two peaks (foreground + background).

show the original image, its histogram, and the thresholded image.

```python
#Load the image
image = data.text()

#Otsu's threshold
threshold, otsu = cv2.threshold(image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

#Display
fig = plt.figure(figsize=(12, 4))

fig.add_subplot(2, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

fig.add_subplot(2, 2, 2)
plt.hist(image.ravel(), bins = 256, range=(0, 256))
plt.title('Original Image Histogram')


fig.add_subplot(2, 2, 3)
plt.imshow(otsu, cmap='gray')
plt.title('Otsu Threshold Image')


fig.add_subplot(2, 2, 4)
plt.hist(otsu.flat, bins = 256, range=(0,256))
plt.title('Otsu Image Histogram')


plt.tight_layout()
plt.show()
```
